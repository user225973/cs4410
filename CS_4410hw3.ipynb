```python
import numpy as np
import matplotlib.pyplot as plt
from sklearn.datasets import load_iris
from sklearn.neighbors import KNeighborsClassifier
from sklearn.model_selection import cross_val_score, GridSearchCV
```

```python
iris = load_iris()
X = iris.data
y = iris.target
```

```python
k_range = range(1, 31)
k_scores = []

for k in k_range:
  knn = KNeighborsClassifier(n_neighbors=k)
  scores = cross_val_score(knn, X, y, cv=10)
  k_scores.append(scores.mean())
```

```python
optml_score = max(k_scores)
optml_k = k_scores.index(optml_score) + 1

print("Optimal K:", optml_k)
print("Best accuracy:", optml_score)
```

```python
plt.figure(figsize=(10, 6))
plt.plot(k_range, k_scores)

plt.xlabel("Value of K for KNN")
plt.ylabel("Cross-Validated Accuracy")
plt.title("Value of K for KNN vs. Cross-Validated Accuracy")

plt.xticks(range(0, 31, 5))
plt.grid(False)
plt.show()
```

```python
knn = KNeighborsClassifier()
grid_search = GridSearchCV(knn, {'n_neighbors': range(1, 31)}, cv=10)
grid_search.fit(X, y)
print("Optimal K:", grid_search.best_params_['n_neighbors'])
```
